```python
import fitz  # PyMuPDF library



def extract_images_from_pdf(pdf_path, scale=4.0):
    doc = fitz.open(pdf_path)
    images = []
    
    for page_num in range(len(doc)):
        page = doc.load_page(page_num)
        
        # Create a transformation matrix to scale the image
        matrix = fitz.Matrix(scale, scale)  # scale the image
        
        # Get a pixmap with the specified scaling
        pix = page.get_pixmap(matrix=matrix,)
        
        # Save each page as an image
        img_path =  img_path0+"\\"f"page_{page_num}.png"  # Adjust this as needed
        pix.save(img_path,)
        images.append(img_path)
    
    return images

# Example usage

images = extract_images_from_pdf(pdf_path)
```
